### 2.1 Tokenizing text using regular expression

```python
# Tokenizing using regular expression
import re

input_text = "Hello world, this is a sample text for LLM."

# split the input text based on whitespaces using regular expression (including whitespaces in list)
result = re.split(r'(\s)', input_text)
print(f"Ouput list based on whitespaces(including whitespace): \n{result}\n")

# split the input text based on whitespaces and punctuations 
result = re.split(r'([,.]|\s)', input_text) # split punctuations mentioned as separate tokens
print(f"Ouput list based on whitespaces & punctuations: \n{result}\n")

# strip whitespace from each item and then filter out any empty strings
result = [item for item in result if item.strip()]
print(f"Ouput list based on whitespaces & punctuations(exclusing whitespaces): \n{result}\n")

# handling more complex input text with punctuatuions
complex_text = "Hello, world, Is this a --text?"
result = re.split(r'([,.:;?_!"()\']|--|\s)', complex_text)
result = [item for item in result if item.strip()]
print(f"Ouput list: \n{result}\n")
```

```python
# tokenizing the dataset using regular expression
processed_text = re.split(r'([,."?_!()"\';:]|--|\s)', raw_text)
processed_text = [item for item in processed_text if item.strip()]
print(f"Generated tokens(first 50): {processed_text[:50]}\n")
print(f"Number of tokens generated: {len(processed_text)}")
```

### 2.2 Generating token IDs using tokenizer

```python
# Tokenizer use a vocabulary which contains a collection of words and their coresponding token ids to generate token ids
# when a new token is given as input, tokenizer finds that token in vocabulary and returns the token id & vice versa

# finding unique tokens from the dataset
unique_tokens = sorted(set(processed_text))
print(f"Unique tokens from dataset: {unique_tokens[:20]}\n")
print(f"Number of the unique tokens present in the dataset: {len(unique_tokens)}\n")

# building a vocabulary which contains <k, v> pairs of str and int
vocabulary = {token:integer for integer, token in enumerate(unique_tokens)}
print(f"Vocabulary generated from dataset: {vocabulary}\n")
print(f"Number of pairs in vocabulary: {len(vocabulary)}\n")
```

```python
# Tokenizer which encodes a token (provide token id for given token) & decodes (provides token for given token id) using vocabulary
class SimpleTokenizerV1:
    def __init__(self, vocabulary):
        self.str_to_int = vocabulary        # creating vocabulary for reference for encoding
        self.int_to_str = {integer:token for token, integer in vocabulary.items()}      # creating a inverse vocabulary for decoding
    
    '''
        converts a token into token ids based on the vocabulary
        input: sentence, vocabulary
        ouput: list of token ids for the input sentence
    '''
    def encode(self, input_text):
        # generating tokens from input sentence
        preprocessed = re.split(r'([,."?_!()"\';:]|--|\s)', input_text)
        preprocessed = [item for item in preprocessed if item.strip()]

        # mapping token ids 
        ouput_ids = [self.str_to_int[token] for token in preprocessed]
        return ouput_ids
    
    '''
        converts a token id into token based on inverse vocabulary generated
        input: list of token ids, inverse vocabulary
        output: list of tokens from vocabulary
    '''
    def decode(self, input_ids):
        tokens = " ".join([self.int_to_str[token_id] for token_id in input_ids])
        output_tokens = re.sub(r'\s+([,.?!"()\'])', r'\1', tokens)      # Replace the whitespaces before specified punctuations
        return output_tokens
```

```python
tokenizer = SimpleTokenizerV1(vocabulary)

sentence = """"It's the last he painted, you know," 
                Mrs. Gisburn said with pardonable pride."""

encoded_text = tokenizer.encode(sentence)
print(f"Token ids generated using tokenizer for \n{sentence}\nis\n{encoded_text}\n")

decoded_text = tokenizer.decode(encoded_text)
print(f"Tokens generated using tokenizer for token list is\n{decoded_text}")
```

### 2.3 Adding special context toekns
- The above simple tokenizer fails if the token is not present in the vocabulary 
- To overcome this we can add special context tokens such as `<|unkown|>` for unkown text and `<|endoftext|>` to denote end of text

```python
# failure of toeknizer
error_sentence = "Hello! This sentence fails in tokenizer"
tokenizer.encode(error_sentence)
```

```python
# adding special context tokens to the dataset
token_list = sorted(set(processed_text))
token_list.extend(["<|unknown|>", "<|endoftext|>"])

# updated vocabulary
vocabulary_V2 = {token:integer for integer, token in enumerate(token_list)}
print(f"Vocabulary generated from dataset V2: {list(vocabulary_V2.items())[-5:]}\n")
print(f"Number of pairs in vocabulary_V2: {len(vocabulary_V2)}\n")
```

```python
# Tokenizer which encodes a token (provide token id for given token) & decodes (provides token for given token id) using vocabulary
class SimpleTokenizerV2:
    def __init__(self, vocabulary):
        self.str_to_int = vocabulary        # creating vocabulary for reference for encoding
        self.int_to_str = {integer:token for token, integer in vocabulary.items()}      # creating a inverse vocabulary for decoding
    
    '''
        converts a token into token ids based on the vocabulary
        input: sentence, vocabulary
        ouput: list of token ids for the input sentence
    '''
    def encode(self, input_text):
        # generating tokens from input sentence
        preprocessed = re.split(r'([,."?_!()"\';:]|--|\s)', input_text)
        preprocessed = [item for item in preprocessed if item.strip()]

        # handling unknown tokens
        preprocessed = [
            token if token in self.str_to_int
            else "<|unknown|>"
            for token in preprocessed
        ]

        # mapping token ids 
        ouput_ids = [self.str_to_int[token] for token in preprocessed]
        return ouput_ids
    
    '''
        converts a token id into token based on inverse vocabulary generated
        input: list of token ids, inverse vocabulary
        output: list of tokens from vocabulary
    '''
    def decode(self, input_ids):
        tokens = " ".join([self.int_to_str[token_id] for token_id in input_ids])
        output_tokens = re.sub(r'\s+([,.?!"()\'])', r'\1', tokens)      # Replace the whitespaces before specified punctuations
        return output_tokens
```

```python
tokenizerV2 = SimpleTokenizerV2(vocabulary_V2)

encoded_text = tokenizerV2.encode(error_sentence)
print(f"Token ids generated using tokenizer v2 for \n{error_sentence}\nis\n{encoded_text}\n")

decoded_text = tokenizerV2.decode(encoded_text)
print(f"Tokens generated using tokenizer v2 for token list is\n{decoded_text}")
```

### 2.5 Data sampling with sliding window
- It will be a computing overhead and performance bottleneck for training LLM if we provide all the dataset input tokens at once(which is not possible if the dataset is huge)
- To make training efficient we need to sample data from dataset for training

```python
# tokenize the dataset which is the-verdict.txt file
encoded_text = gpt2_tokenizer.encode(raw_text, allowed_special={'<|endoftext|>'})

print(f"Length of tokens generated from dataset is: {len(encoded_text)}")
```

```python
# we take a context window of size 'n' and create chunks from dataset token ids into inputs and targets as following example
# LLM predicts one word at a time thus for a given set of input words it predicts the next word(target) 

encoded_sample = encoded_text[:50]      # taking first 50 tokens for illustration
context_window = 4

x = encoded_sample[:context_window]         # sample input chunk
y = encoded_sample[1:context_window+1]      # sample target chunk which contain the next word in the sequence of dataset

print(f"input  x: {x}\n")
print(f"target y:\t{y}")
```

```python
# One word by another, the prediction would look like as follows:
for i in range(1, context_window+1):
    context = encoded_sample[:i]
    label = encoded_sample[i]

    # token id illustration
    print(f"{context} --------> {label}")
    # word illustration
    print(f"{gpt2_tokenizer.decode(context)} --------> {gpt2_tokenizer.decode([label])}\n")
```

- pytorch can be used to implement a simple data loader that iterates over the input dataset and returns the inputs and targets shifted by one

To use GPU, install pytorch compatible with CUDA version present in the system
- use `nvidia-smi` to fins the CUDA version in the system
- use [pytorch website](https://pytorch.org/) to get the pip command

```python
import torch
print(f"pyTorch version installed: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")
```

```python
# Build dataset and dataloader for training the LLM
from torch.utils.data import Dataset, DataLoader

class GPTDatasetV1(Dataset):
    def __init__(self, input_txt, tokenizer, max_length, stride):
        self.input_ids = []
        self.target_ids = []
        
        # tokenize the input sentence
        token_ids = tokenizer.encode(input_txt, allowed_special={"<|endoftext|>"})

        # use sliding window to create the input and target chunks
        for i in range(0, len(token_ids) - max_length, stride):
            input_chunk = token_ids[i : i+max_length]
            target_chunk = token_ids[i+1 : i+max_length+1]
            self.input_ids.append(torch.tensor(input_chunk))
            self.target_ids.append(torch.tensor(target_chunk))

    def __getitem__(self, index):
        return self.input_ids[index], self.target_ids[index]
    
    def __len__(self):
        return len(self.input_ids)
```

```python
def create_dataloader_V1(input_text, batch_size=4, max_length=256, 
                         stride=128, shuffle=True, drop_last=True, num_workers=0):
    # Initialize tokenizer
    tokenizer = tiktoken.get_encoding("gpt2")

    # create dataset
    train_dataset = GPTDatasetV1(input_text, tokenizer, max_length, stride)

    # create dataloader
    dataloader = DataLoader(
        dataset = train_dataset,
        batch_size = batch_size,
        shuffle = shuffle,
        drop_last = drop_last,
        num_workers = num_workers
    )

    return dataloader
```

```python
# create a dataloader using input dataset "the-verdict.txt"
# dataloader 01 with max_length 4 and stride 1 (i.e) sliding window is moved one token for each input, target pair
train_dataloader = create_dataloader_V1(input_text=raw_text, batch_size=1, max_length=4, stride=1, shuffle=False)

# iterate over dataloader to sample 
train_dataloader_iterator = iter(train_dataloader)
first_bacth = next(train_dataloader_iterator)
print(f"The first batch is:\n{first_bacth}")

second_batch = next(train_dataloader_iterator)
print(f"The second_batch is:\n{second_batch}")
```

```python
# the previous example can excessively repeat tokens making the training for LLM inefficient if the stride is less and max_length is high
# dataloader 02 with max_length 4 and stride 4 (i.e) sliding window is moved four token for each input, target pair
train_dataloader_02 = create_dataloader_V1(input_text=raw_text, batch_size=1, max_length=4, stride=4, shuffle=False)

# iterate over dataloader to sample 
train_dataloader_iterator_02 = iter(train_dataloader_02)
first_bacth = next(train_dataloader_iterator_02)
print(f"The first batch is:\n{first_bacth}")

second_batch = next(train_dataloader_iterator_02)
print(f"The second_batch is:\n{second_batch}")
```

#### 2.6 Creating token embeddings
- converting token ids into embeddings of n dimension for efficient floating point arithmetic using embedding layer
- The embedding layer is usually a part of LLM which has weight parameters that are trained in training
- Embedding layer is equivalent to liner layer while the input is given as one-hot encoding for linear layer
- output of embedding layer is equal to one hot encoding with matrix multiplication
- It has a weight matrix of *vocabulary_size x output_dimension* (i.e) one tensor of size *1 x output_dimension* for each token in vocabulary size

```python
# conside an example which has following token ids after tokenization
input_ids = torch.tensor([2,3,5,1])

# consider a small vocabulary size of 6 <word, token> pairs and embedding layer output as 3
vocab_size = 6
output_dim = 3

# embedding layer with the specifications
embedding_layer = torch.nn.Embedding(vocab_size, output_dim)

# this results in a weight matrix (vocab_size x output_dim) which are the parameters of embedding layer
print(f"The weight matrix of embedding layer is \n{embedding_layer.weight}\nwith size ({embedding_layer.weight.shape})")
```

```python
# the output embedding for an input token id is given as follows
print(embedding_layer(torch.tensor([3])))       # provides the weight matrix row where index - input id
```

```python
# output for the input id tensor
print(embedding_layer(input_ids))
```

```python
# GPT 2 embedding layer 
vocab_size = 50257
output_dim = 256

gpt2_embedding_layer = torch.nn.Embedding(vocab_size, output_dim)
```

```python
# create a dataloader with batch_size 8 
max_length = 4

gpt2_dataloader = create_dataloader_V1(
    input_text= raw_text,
    batch_size= 8,
    stride= max_length,
    max_length= max_length,
    shuffle= False
)

data_iter = iter(gpt2_dataloader)
inputs, targets = next(data_iter)
```

```python
print(f"Token IDs: \n{inputs}")
print(f"Shape of input token matrix: {inputs.shape}")
```

```python
gpt2_token_embeddings = gpt2_embedding_layer(inputs)
print(f"The output of embedding layer is \n{gpt2_token_embeddings}\nwith size ({gpt2_token_embeddings.shape})")
```

- embedding layer produces identical embeddings for a given token id regardless of their position in the input sentence
- to incorporate positional data into the input, positional embeddings are used 
- positional embeddings is another layer which adds a fixed value taken from weight matrix of that layer to the input id 

```python
# GPT2 uses absolute position embeddings whcih can be implemented by adding another embedding layer
context_length = max_length

positional_embedding_layer = torch.nn.Embedding(context_length, output_dim)
print(f"The weight matrix of positional embedding layer is \n{positional_embedding_layer.weight}\nwith size ({positional_embedding_layer.weight.shape})")
```

```python
# the context window size of GPT2 is 1024 hence the postional embedding layer will have shape of 1024 x 256 
pos_embeddings = positional_embedding_layer(torch.arange(max_length))
print(f"The output of positional embedding layer is \n{pos_embeddings}\nwith size ({pos_embeddings.shape})")
```

```python
input_embeddings = gpt2_token_embeddings + pos_embeddings
print(f"The output of embedding layers is \n{input_embeddings}\nwith size ({input_embeddings.shape})")
```
